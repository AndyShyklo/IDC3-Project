```python
import sys
print(sys.executable)

import geopandas as gpd
import pandas as pd
from pathlib import Path

BASE_DIR = Path.cwd().resolve()

uhfDis = gpd.read_file(BASE_DIR/"UHF42.geo.json").to_crs(epsg=2263)
# zones = gpd.read_file(BASE_DIR/"zones.json").to_crs(epsg=2263)

shapefile_path = BASE_DIR / "nyzd.shp"

zones = gpd.read_file(shapefile_path).to_crs(epsg=2263)

uhfDis["total_zone_sqft"] = uhfDis.geometry.area

result = uhfDis.copy()


# All M zones
is_m = zones["ZONEDIST"].str.match(r"^M[1-3]")
is_mixed = zones["ZONEDIST"].str.contains("/") | zones["ZONEDIST"].str.contains("MX")
pure_m = zones[is_m & ~is_mixed].copy()

# print(pure_m)

pure_m_dissolved = pure_m.dissolve()

intersection = gpd.overlay(uhfDis, pure_m_dissolved, how="intersection")

intersection["total_m_sqft"] = intersection.geometry.area

uhf_col = "GEOCODE" if "GEOCODE" in intersection.columns else "id"
m_totals = intersection.groupby(uhf_col)["total_m_sqft"].sum().reset_index()

result = result.merge(m_totals, on=uhf_col, how="left")
result["total_m_sqft"] = result["total_m_sqft"].fillna(0)
result["percent_total_m_zone"] = (result["total_m_sqft"] / result["total_zone_sqft"]) * 100



# M3 zones
is_m = zones["ZONEDIST"].str.match(r"^M3")
is_mixed = zones["ZONEDIST"].str.contains("/") | zones["ZONEDIST"].str.contains("MX")
pure_m = zones[is_m & ~is_mixed].copy()

pure_m_dissolved = pure_m.dissolve()

intersection = gpd.overlay(uhfDis, pure_m_dissolved, how="intersection")

intersection["m3_sqft"] = intersection.geometry.area

uhf_col = "GEOCODE" if "GEOCODE" in intersection.columns else "id"
m_totals = intersection.groupby(uhf_col)["m3_sqft"].sum().reset_index()

result = result.merge(m_totals, on=uhf_col, how="left")
result["m3_sqft"] = result["m3_sqft"].fillna(0)
result["percent_m3_zone"] = (result["m3_sqft"] / result["total_zone_sqft"]) * 100



# M2 zones
is_m = zones["ZONEDIST"].str.match(r"^M2")
is_mixed = zones["ZONEDIST"].str.contains("/") | zones["ZONEDIST"].str.contains("MX")
pure_m = zones[is_m & ~is_mixed].copy()

pure_m_dissolved = pure_m.dissolve()

intersection = gpd.overlay(uhfDis, pure_m_dissolved, how="intersection")

intersection["m2_sqft"] = intersection.geometry.area

uhf_col = "GEOCODE" if "GEOCODE" in intersection.columns else "id"
m_totals = intersection.groupby(uhf_col)["m2_sqft"].sum().reset_index()

result = result.merge(m_totals, on=uhf_col, how="left")
result["m2_sqft"] = result["m2_sqft"].fillna(0)
result["percent_m2_zone"] = (result["m2_sqft"] / result["total_zone_sqft"]) * 100



# M1 zones
is_m = zones["ZONEDIST"].str.match(r"^M1")
is_mixed = zones["ZONEDIST"].str.contains("/") | zones["ZONEDIST"].str.contains("MX")
pure_m = zones[is_m & ~is_mixed].copy()

pure_m_dissolved = pure_m.dissolve()

intersection = gpd.overlay(uhfDis, pure_m_dissolved, how="intersection")

intersection["m1_sqft"] = intersection.geometry.area

uhf_col = "GEOCODE" if "GEOCODE" in intersection.columns else "id"
m_totals = intersection.groupby(uhf_col)["m1_sqft"].sum().reset_index()

result = result.merge(m_totals, on=uhf_col, how="left")
result["m1_sqft"] = result["m1_sqft"].fillna(0)
result["percent_m1_zone"] = (result["m1_sqft"] / result["total_zone_sqft"]) * 100



# C8 zones
is_m = zones["ZONEDIST"].str.match(r"^C8")
is_mixed = zones["ZONEDIST"].str.contains("/") | zones["ZONEDIST"].str.contains("MX")
pure_m = zones[is_m & ~is_mixed].copy()

pure_m_dissolved = pure_m.dissolve()

intersection = gpd.overlay(uhfDis, pure_m_dissolved, how="intersection")

intersection["c8_sqft"] = intersection.geometry.area

uhf_col = "GEOCODE" if "GEOCODE" in intersection.columns else "id"
m_totals = intersection.groupby(uhf_col)["c8_sqft"].sum().reset_index()

result = result.merge(m_totals, on=uhf_col, how="left")
result["c8_sqft"] = result["c8_sqft"].fillna(0)
result["percent_c8_zone"] = (result["c8_sqft"] / result["total_zone_sqft"]) * 100



def printAll():
    output_cols = [uhf_col, "GEONAME", "BOROUGH", "AsthmaRate", "PovertyRate", "total_zone_sqft", "total_m_sqft", "m3_sqft", "m2_sqft", "m1_sqft", "c8_sqft", "percent_total_m_zone", "percent_m3_zone", "percent_m2_zone", "percent_m1_zone", "percent_c8_zone"]
    final_table = result[[c for c in output_cols if c in result.columns]].sort_values(
        by="AsthmaRate", ascending=False
    )
    #print(final_table)
    final_table.to_csv("m_table_analysis.csv", index=False)

# old ver, doesn't print full
def printMain():
    output_cols = ["GEONAME", "BOROUGH", "AsthmaRate", "percent_m_zone"]
    final_table = result[[c for c in output_cols if c in result.columns]].sort_values(
        by="AsthmaRate", ascending=False
    )
    print(final_table)

printAll()
# printMain()


# for line in uhfDis:
#     gpd.overlay(line, )

# print(uhfDis)
```
